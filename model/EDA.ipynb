```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd
```

```python
df = pd.read_csv("data/diabetes_binary_health_indicators_BRFSS2015.csv")
df.head()
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
print("Missing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
print(df["Diabetes_binary"].value_counts())

print("\nPercentage:")
print(df["Diabetes_binary"].value_counts(normalize=True) * 100)
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="Diabetes_binary")

plt.title("Diabetes Class Distribution")
plt.xlabel("Diabetes Binary")
plt.ylabel("Number of People")
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15,4))

sns.histplot(df["BMI"], bins=30, ax=axes[0])
axes[0].set_title("BMI Distribution")

sns.histplot(df["Age"], bins=13, ax=axes[1])
axes[1].set_title("Age Distribution")

sns.countplot(data=df, x="GenHlth", hue="Diabetes_binary", ax=axes[2])
axes[2].set_title("General Health vs Diabetes")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14,10))

corr = df.corr(numeric_only=True)

sns.heatmap(corr, cmap="coolwarm", center=0)

plt.title("Feature Correlation Heatmap")
plt.show()
```
